```python
import tensorflow as tf
from tensorflow import keras 
from keras.models import Sequential
from keras.layers import Flatten, Dense
from keras.applications import MobileNetV2
```

```python
plants = ['Apple', 'Bell Pepper', 'Cherry', 'Corn', 'Cucumber', 'Grape', 'Guava', 'Lemon', 'Mango', 'Peach', 'Potato', 'Rice', 'Strawberry', 'Sugarcane', 'Tea', 'Tomato', 'Wheat']
```

```python
import os

disease_models = {}
callbacks = [
    keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
]

for plant in plants:
    num_classes = len(list(os.listdir(f'./Dataset/Leaf/{plant}')))
    print(f'Training model for {plant}')
    dataset = keras.preprocessing.image_dataset_from_directory(
        directory=f'./Dataset/Leaf/{plant}',
        shuffle=True,
        batch_size=16,
        image_size=(128,128)
    )
    train_ds = dataset.take(int(len(dataset)*0.7)).cache().shuffle(1000).prefetch(buffer_size=tf.data.experimental.AUTOTUNE)
    valid_ds = dataset.skip(int(len(dataset)*0.7)).take(int(len(dataset)*0.2)).cache().prefetch(buffer_size=tf.data.experimental.AUTOTUNE)
    test_ds = dataset.skip(int(len(dataset)*0.9)).cache().prefetch(buffer_size=tf.data.experimental.AUTOTUNE)

    conv_base = MobileNetV2(weights='imagenet', include_top=False, input_shape=(128,128,3))
    model = Sequential()
    model.add(conv_base)
    model.add(Flatten())
    model.add(Dense(num_classes, activation='softmax'))

    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    model.fit(train_ds, validation_data=valid_ds, epochs=50, callbacks=callbacks)

    disease_models[plant] = model
    model.save(f'../saved_models/diseases/{plant}.keras')
    print(f'Model for {plant} saved')
```
